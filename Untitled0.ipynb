# Satellite Orbit Explorer · built during Space Career Week 2026 · GNSS Academy

This educational notebook explores satellite orbits using a two-body Keplerian approximation with a simple Greenwich Earth rotation model. We will build this in 8 stages:

1. **Setup**: Imports, constants, directories, and step configuration.
2. **Load GP JSON**: Fetch or load satellite General Perturbations (GP) datasets.
3. **Parse Elements**: Convert SGP4 mean elements from GP JSON to Keplerian parameters.
4. **Time Window**: Configure propagation timelines.
5. **Propagate**: Compute inertial and Earth-fixed coordinates over time.
6. **Orbit Summary**: Generate statistics like altitude, period, and ground speed.
7. **Plots**: Visualize the orbit path and ground tracks.
8. **KML and Report**: Export visual tracks and complete the summary report.

## Stage 2: Load GP JSON

OMM (Orbit Mean-elements Message) JSON format is vastly superior to the legacy two-line element (TLE) format because it provides descriptive, named key-value fields instead of rigid column-based parameters. Furthermore, it represents dates using modern, standard ISO-8601 strings and natively supports modern six-digit NORAD catalog numbers without truncation issues.

```python
def extract_inputs():
    """Ensures raw data files are extracted into IN_DIR, waiting for the zip if missing."""
    required_files = ['orbital_elements.json', 'data_manifest.json', 'world_map.txt']
    all_exist = all((IN_DIR / f).is_file() for f in required_files)

    if all_exist:
        print("All required files are already present in IN_DIR.")
        return

    print("--- Input Files Not Found in IN_DIR ---")
    print("Please drag 'orbit_workshop_inputs.zip' into the main /content folder in Colab.")
    print("Waiting for file upload...")

    start_time = time.time()
    timeout = 15 * 60  # 15 minutes
    zip_path = None

    while True:
        if time.time() - start_time > timeout:
            raise TimeoutError("Upload timed out after 15 minutes.")

        # Search for a matching zip file in /content
        candidates = list(Path('/content').glob('orbit_workshop_inputs*.zip'))
        if candidates:
            candidate = candidates[0]
            if zipfile.is_zipfile(candidate):
                zip_path = candidate
                break

        time.sleep(2)

    print(f"Found upload: {zip_path.name}. Extracting contents...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        for f in required_files:
            # Extract only the three specified files
            try:
                zip_ref.extract(f, path=IN_DIR)
                print(f"  Extracted: {f}")
            except KeyError:
                raise FileNotFoundError(f"{f} not found inside the zip file.")

def read_gp_json(path):
    """Loads GP elements JSON and validates the presence of critical OMM fields."""
    if not path.is_file():
        raise FileNotFoundError(f"JSON file not found: {path}")

    with open(path, 'r') as f:
        records = json.load(f)

    required_fields = [
        'OBJECT_NAME', 'NORAD_CAT_ID', 'EPOCH', 'MEAN_MOTION',
        'ECCENTRICITY', 'INCLINATION', 'RA_OF_ASC_NODE',
        'ARG_OF_PERICENTER', 'MEAN_ANOMALY'
    ]

    for idx, rec in enumerate(records):
        missing = [field for field in required_fields if field not in rec]
        if missing:
            raise ValueError(f"Record at index {idx} ({rec.get('OBJECT_NAME', 'Unknown')}) is missing keys: {missing}")

    return records

# Execute Stage 2 pipeline
extract_inputs()

sats = read_gp_json(IN_DIR / 'orbital_elements.json')

with open(IN_DIR / 'data_manifest.json', 'r') as f:
    data_manifest = json.load(f)

source_note = f"Data source: {data_manifest.get('description', 'CelesTrak GP Dataset')} (Retrieved: {data_manifest['retrieved_utc']})"

print(f"\n{source_note}")
print(f"Loaded {len(sats)} satellites successfully:\n")
for sat in sats:
    print(f"Name: {sat['OBJECT_NAME']:<20} | NORAD ID: {sat['NORAD_CAT_ID']:<6} | Epoch: {sat['EPOCH']}")
```

## Stage 3: Parse Elements

The table below maps the GP JSON fields to their classical Keplerian symbols and unit conversions used in our two-body model:

| GP JSON Field | Keplerian Symbol | OMM Native Unit | Model Working Unit | Conversion / Formula |
|---|---|---|---|---|
| `MEAN_MOTION` | $n$ | Revolutions / Day | Radians / Second | $n_{\text{rad/s}} = \frac{2\pi \cdot \text{MEAN\_MOTION}}{86400}$ |
| Calculated | $a$ | N/A (Derived) | Kilometers (km) | $a = \left(\frac{\mu}{n^2}\right)^{1/3}$ |
| `ECCENTRICITY` | $e$ | Decimal | Decimal | None (used directly) |
| `INCLINATION` | $i$ | Degrees | Radians | $\text{rad} = \text{deg} \cdot \frac{\pi}{180}$ |
| `RA_OF_ASC_NODE` | $\Omega$ | Degrees | Radians | $\text{rad} = \text{deg} \cdot \frac{\pi}{180}$ |
| `ARG_OF_PERICENTER`| $\omega$ | Degrees | Radians | $\text{rad} = \text{deg} \cdot \frac{\pi}{180}$ |
| `MEAN_ANOMALY` | $M_0$ | Degrees | Radians | $\text{rad} = \text{deg} \cdot \frac{\pi}{180}$ |

> **Note:** The input parameters from CelesTrak are SGP4 *mean elements*, which are mathematically designed to average out various perturbing forces (like Earth's oblateness, $J_2$). Using them directly in an unperturbed two-body Keplerian propagator is an educational approximation, not a precision orbit prediction model.

```python
def parse_gp_epoch(text):
    """Parses a CelesTrak ISO-8601 epoch string as a UTC-aware datetime."""
    # Strip trailing 'Z' if present, and force tzinfo to UTC
    clean_text = text.rstrip('Z')
    return datetime.fromisoformat(clean_text).replace(tzinfo=timezone.utc)

def parse_gp_elements(rec):
    """Converts raw GP record fields to physical Keplerian elements in model units."""
    name = rec['OBJECT_NAME']
    norad_id = int(rec['NORAD_CAT_ID'])

    epoch = parse_gp_epoch(rec['EPOCH'])

    # Mean motion: rev/day to rad/s
    mean_motion_rev_day = rec['MEAN_MOTION']
    n_rad_s = (2.0 * math.pi * mean_motion_rev_day) / 86400.0

    # Semi-major axis (a) from Kepler's third law: a^3 = MU / n^2
    a_km = (MU_EARTH_KM3_S2 / (n_rad_s ** 2)) ** (1.0 / 3.0)

    # Eccentricity is already parsed as decimal fraction
    e = rec['ECCENTRICITY']

    # Convert angular values from degrees to radians
    i_rad = math.radians(rec['INCLINATION'])
    raan_rad = math.radians(rec['RA_OF_ASC_NODE'])
    argp_rad = math.radians(rec['ARG_OF_PERICENTER'])
    M0_rad = math.radians(rec['MEAN_ANOMALY'])

    return {
        'name': name,
        'norad_id': norad_id,
        'label': f"{name} [{norad_id}]",
        'epoch': epoch,
        'n_rad_s': n_rad_s,
        'a_km': a_km,
        'e': e,
        'i_rad': i_rad,
        'raan_rad': raan_rad,
        'argp_rad': argp_rad,
        'M0_rad': M0_rad,
        'i_deg': rec['INCLINATION'],
        'raan_deg': rec['RA_OF_ASC_NODE'],
        'argp_deg': rec['ARG_OF_PERICENTER'],
        'M0_deg': rec['MEAN_ANOMALY']
    }

def elements_table(elements_list):
    """Generates formated text table lines summarizing parsed Keplerian elements."""
    lines = []
    header = (f"{'Satellite Name':<25} | {'NORAD':<5} | {'Epoch (UTC)':<26} | "
              f"{'a (km)':<9} | {'e':<8} | {'i (deg)':<7} | {'Ω (deg)':<7} | {'ω (deg)':<7} | {'M0 (deg)':<7}")
    separator = "-" * len(header)

    lines.append(header)
    lines.append(separator)

    for el in elements_list:
        epoch_str = el['epoch'].strftime('%Y-%m-%d %H:%M:%S.%f')[:-3]
        row = (f"{el['name']:<25} | {el['norad_id']:<5} | {epoch_str:<26} | "
               f"{el['a_km']:9.2f} | {el['e']:8.6f} | {el['i_deg']:7.2f} | {el['raan_deg']:7.2f} | {el['argp_deg']:7.2f} | {el['M0_deg']:7.2f}")
        lines.append(row)

    return lines

# Ensure sats list exists from Stage 2
if 'sats' not in globals():
    raise ValueError("Stage 2 output 'sats' is missing. Please run Stage 2 first.")

# Convert GP records to elements list
elements = [parse_gp_elements(s) for s in sats]

# Generate and print the summary table
table_lines = elements_table(elements)
for line in table_lines:
    print(line)
```

## Stage 4: Time Window and Propagation Grid

Having a shared time grid allows us to directly compare satellite positions and plot them relative to a synchronized global clock. However, because each satellite's GP record is propagated from its own unique historical `epoch`, the mathematical error in our simple model grows larger as the time offset between that epoch and the propagation window increases. For example, our dataset's QZS-2 record dates back to October 17, 2026—making it significantly older than the other November 2026 records and susceptible to greater error.

```python
# Ensure elements exist from earlier stages
if 'elements' not in globals():
    raise ValueError("Stage 3 output 'elements' is missing. Please run Stage 3 first.")

# 1. Determine the newest epoch across all satellites
newest_epoch = max(el['epoch'] for el in elements)

# 2. Floor this newest epoch to UTC midnight of the same day
start_time = newest_epoch.replace(hour=0, minute=0, second=0, microsecond=0, tzinfo=timezone.utc)
end_time = start_time + timedelta(hours=DURATION_HOURS)

# 3. Create the common UTC time grid
total_seconds = int((end_time - start_time).total_seconds())
n_points = total_seconds // STEP_SECONDS + 1
times = [start_time + timedelta(seconds=i * STEP_SECONDS) for i in range(n_points)]

# Print time window metrics
print("=== Stage 4 Time Window Configured ===")
print(f"Newest Epoch:      {newest_epoch}")
print(f"Common Start (UTC): {start_time}")
print(f"Common End (UTC):   {end_time}")
print(f"Total Grid Points:  {n_points} points (Step: {STEP_SECONDS} s)\n")

# 4. Assess epoch offsets and flag stale elements (> 7 days)
print(f"{'Satellite Name':<25} | {'Epoch Offset (days)':<19} | {'Status':<10}")
print("-" * 60)
for el in elements:
    offset_days = (start_time - el['epoch']).total_seconds() / 86400.0
    is_stale = offset_days > 7.0
    status = "STALE RECORD" if is_stale else "FRESH"
    print(f"{el['name']:<25} | {offset_days:19.4f} | {status:<10}")
```

## Stage 5: Propagate Orbits

To compute the coordinates of our satellites over our time grid, we implement a classic two-body unperturbed Keplerian propagator. The propagation operates in seven distinct steps:

1. **Mean Anomaly Update**: Project the Mean Anomaly $M$ forward in time from the initial epoch using the Mean Motion $n$: $M = M_0 + n(t - t_0)$.
2. **Kepler's Equation Solver**: Solve the transcendental equation $M = E - e\sin(E)$ using the iterative Newton-Raphson method to find the Eccentric Anomaly $E$.
3. **True Anomaly and Position in Orbital Plane**: Compute the radial distance $r = a(1 - e\cos(E))$ and the True Anomaly $\nu = \text{atan2}(\sqrt{1-e^2}\sin(E), \cos(E)-e)$. The argument of latitude is $u = \omega + \nu$.
4. **Inertial Geocentric (ECI) Frame**: Project the position onto the geocentric equatorial inertial frame using the inclination $i$ and Right Ascension of the Ascending Node (RAAN, $\Omega$).
5. **Greenwich Rotation**: Rotate the inertial $X/Y$ coordinates to the Earth-Centered Earth-Fixed (ECEF) frame using the Greenwich Mean Sidereal Time (GMST) angle $\theta$ calculated for each epoch.
6. **Geodetic Spherical Coordinates**: Extract Latitude, Longitude, and Altitude from the ECEF position vectors under a spherical Earth model.
7. **Vis-Viva Velocity**: Compute orbital speed using the energy conservation equation $v = \sqrt{\mu \left(\frac{2}{r} - \frac{1}{a}\right)}$.

```python
# Ensure required elements exist
if 'elements' not in globals() or 'times' not in globals():
    raise ValueError("Stage 3 and 4 variables must be available in the global scope.")

def solve_kepler(M, e, tol=1e-12, max_iter=50):
    """Solves Kepler's Equation M = E - e*sin(E) using Newton-Raphson iteration."""
    # Start close to the solution
    E = math.pi if e > 0.8 else M
    for _ in range(max_iter):
        f = E - e * math.sin(E) - M
        f_prime = 1.0 - e * math.cos(E)
        delta = f / f_prime
        E -= delta
        if abs(delta) < tol:
            return E
    raise RuntimeError(f"Kepler solver failed to converge for M={M}, e={e} within {max_iter} iterations.")

def gmst_rad(t):
    """Computes Greenwich Mean Sidereal Time (GMST) in radians using IAU-1982 formulation."""
    # J2000 epoch is 2000-01-01 12:00:00 UTC
    j2000 = datetime(2000, 1, 1, 12, 0, 0, tzinfo=timezone.utc)
    delta_time = t - j2000
    d = delta_time.total_seconds() / 86400.0
    T = d / 36525.0

    # GMST expression in degrees (IAU 1982)
    theta_deg = 280.46061837 + 360.98564736629 * d + 0.000387933 * (T ** 2) - (T ** 3) / 38710000.0
    return math.radians(theta_deg % 360.0)

def propagate(el, times):
    """Propagates Keplerian elements across a list of datetime objects."""
    track = []
    epoch = el['epoch']
    n = el['n_rad_s']
    a = el['a_km']
    e = el['e']
    i = el['i_rad']
    raan = el['raan_rad']
    argp = el['argp_rad']
    M0 = el['M0_rad']

    for t in times:
        # 1. Update Mean Anomaly M modulo 2*pi
        dt = (t - epoch).total_seconds()
        M = (M0 + n * dt) % (2.0 * math.pi)
        if M < 0:
            M += 2.0 * math.pi

        # 2. Solve Kepler's Equation for E
        E = solve_kepler(M, e)

        # 3. Radial distance & True Anomaly in orbital plane
        r = a * (1.0 - e * math.cos(E))
        sin_nu = (math.sqrt(1.0 - e ** 2) * math.sin(E))
        cos_nu = (math.cos(E) - e)
        nu = math.atan2(sin_nu, cos_nu)
        u = argp + nu

        # 4. Inertial positions (ECI)
        cos_u, sin_u = math.cos(u), math.sin(u)
        cos_raan, sin_raan = math.cos(raan), math.sin(raan)
        cos_i, sin_i = math.cos(i), math.sin(i)

        x = r * (cos_raan * cos_u - sin_raan * sin_u * cos_i)
        y = r * (sin_raan * cos_u + cos_raan * sin_u * cos_i)
        z = r * sin_u * sin_i

        # 5. ECEF rotation by GMST angle
        theta = gmst_rad(t)
        cos_theta, sin_theta = math.cos(theta), math.sin(theta)
        xp = cos_theta * x + sin_theta * y
        yp = -sin_theta * x + cos_theta * y

        # 6. Lat, Lon (degrees) and Altitude
        lat_deg = math.degrees(math.atan2(z, math.hypot(xp, yp)))
        lon_deg = math.degrees(math.atan2(yp, xp))
        alt_km = r - EARTH_RADIUS_KM

        # 7. Vis-viva speed
        speed_km_s = math.sqrt(MU_EARTH_KM3_S2 * (2.0 / r - 1.0 / a))

        track.append({
            'time': t,
            'lat_deg': lat_deg,
            'lon_deg': lon_deg,
            'alt_km': alt_km,
            'speed_km_s': speed_km_s
        })
    return track

# --- Mathematical validation checks using asserts ---
# Validate GMST at J2000 is approximately 280.46 degrees
j2000_check = datetime(2000, 1, 1, 12, 0, 0, tzinfo=timezone.utc)
assert math.isclose(math.degrees(gmst_rad(j2000_check)), 280.46061837, abs_tol=1e-5), "GMST J2000 check failed"

# Validate solve_kepler numerical convergence
for test_M in [0.0, 1.0, math.pi, 5.0]:
    for test_e in [0.0, 0.1, 0.5]:
        test_E = solve_kepler(test_M, test_e)
        residual = abs(test_E - test_e * math.sin(test_E) - test_M) % (2.0 * math.pi)
        if residual > math.pi:
            residual = 2.0 * math.pi - residual
        assert residual < 1e-10, f"Kepler solver residual check failed for M={test_M}, e={test_e}"

# Propagate all tracks
tracks = [propagate(el, times) for el in elements]

# Validate geostationary behavior of SES-5 (NORAD 38652) if present
ses5_track = next((t for el, t in zip(elements, tracks) if el['norad_id'] == 38652), None)
if ses5_track:
    for pt in ses5_track:
        # Geostationary longitude of SES-5 should hover around 5.0 East (+/- 1.0 deg margin)
        assert 4.0 <= pt['lon_deg'] <= 6.0, f"SES-5 geostationary check failed: found {pt['lon_deg']} deg"
    print("Geostationary validation check: Passed (SES-5 stays within 5.0° E +/- 1°)")

print("=== Stage 5 Verification: Orbits Propagated Successfully ===\n")

# Print sample points
for el, track in zip(elements, tracks):
    p_first, p_mid, p_last = track[0], track[len(track)//2], track[-1]
    print(f"Satellite: {el['label']}")
    print(f"  First Point: Lat = {p_first['lat_deg']:7.2f}°, Lon = {p_first['lon_deg']:7.2f}°, Alt = {p_first['alt_km']:8.1f} km, Speed = {p_first['speed_km_s']:5.3f} km/s")
    print(f"  Middle Point:Lat = {p_mid['lat_deg']:7.2f}°, Lon = {p_mid['lon_deg']:7.2f}°, Alt = {p_mid['alt_km']:8.1f} km, Speed = {p_mid['speed_km_s']:5.3f} km/s")
    print(f"  Last Point:  Lat = {p_last['lat_deg']:7.2f}°, Lon = {p_last['lon_deg']:7.2f}°, Alt = {p_last['alt_km']:8.1f} km, Speed = {p_last['speed_km_s']:5.3f} km/s\n")
```

## Stage 6: Orbit Summary

By analyzing our simulated telemetry data, we can profile each satellite's flight characteristics. Based on their altitudes and geometries, they fall into distinct orbital regimes:
* **PULSAR-0** is a Low Earth Orbit (LEO) satellite, characterized by a fast speed, low altitude, and short period.
* **GSAT0225** resides in Medium Earth Orbit (MEO) with a period of roughly 14 hours.
* **SES-5** is Geostationary (GEO), staying locked over 5.0° East with near-zero relative movement.
* **QZS-2** is an eccentric Highly Elliptical Orbit (HEO) designed to dwell over Japan and East Asia.

```python
# Ensure elements and tracks exist from earlier stages
if 'elements' not in globals() or 'tracks' not in globals():
    raise ValueError("Stage 3 and 5 outputs ('elements', 'tracks') are missing.")

def compute_summary_table(elements_list, tracks_list):
    """Computes key metrics for each satellite and builds a text summary table."""
    lines = []
    header = (f"{'Satellite Name':<25} | {'NORAD':<5} | "
              f"{'Min Alt (km)':<12} | {'Max Alt (km)':<12} | {'Avg Alt (km)':<12} | "
              f"{'Period (h)':<10} | {'Speed Range (km/s)':<19}")
    separator = "-" * len(header)

    lines.append(header)
    lines.append(separator)

    for el, track in zip(elements_list, tracks_list):
        alts = [pt['alt_km'] for pt in track]
        speeds = [pt['speed_km_s'] for pt in track]

        min_alt = min(alts)
        max_alt = max(alts)
        avg_alt = sum(alts) / len(alts)

        min_speed = min(speeds)
        max_speed = max(speeds)

        # Compute period: T = 2*pi / n, convert from seconds to hours
        period_hours = (2.0 * math.pi / el['n_rad_s']) / 3600.0

        row = (f"{el['name']:<25} | {el['norad_id']:<5} | "
               f"{min_alt:12.1f} | {max_alt:12.1f} | {avg_alt:12.1f} | "
               f"{period_hours:10.2f} | {min_speed:5.3f} - {max_speed:5.3f}")
        lines.append(row)

    return lines

# Calculate and print the summary report
summary_lines = compute_summary_table(elements, tracks)
print("=== Stage 6: Orbit Analysis Summary ===")
for line in summary_lines:
    print(line)
print("========================================")
```

## Stage 7: Plots

In this stage, we construct the spatial and physical visualizations of our satellite tracks:
1. **Ground Tracks (Figure 1)**: Visualizes the geodetic latitudes and longitudes over a global plate carrée projection map. We parse the discrete coastline segments from `world_map.txt` and plot our satellite ground paths as discrete scatter plots. This avoids drawing incorrect wrapping lines whenever a satellite crosses the +/-180° international date line.
2. **Velocity Profiles (Figure 2)**: Visualizes the orbital velocities ($v$) in km/s plotted over time as a vertical stack of subplots sharing a common timeline axis.

```python
import numpy as np

# Ensure inputs from previous stages are present
if 'elements' not in globals() or 'tracks' not in globals():
    raise ValueError("Stage 3 and 5 outputs ('elements', 'tracks') are missing.")

def load_coastline(path):
    """Parses world_map.txt which has 'latitude longitude' per line, separated by blank lines."""
    if not path.is_file():
        raise FileNotFoundError(f"Coastline map file not found at: {path}")

    segments = []
    current_segment = []

    with open(path, 'r') as f:
        for line in f:
            line = line.strip()
            if not line:
                if current_segment:
                    segments.append(current_segment)
                    current_segment = []
            else:
                # Latitude is first, longitude is second
                parts = line.split()
                if len(parts) == 2:
                    lat, lon = float(parts[0]), float(parts[1])
                    current_segment.append((lon, lat))

    if current_segment:
        segments.append(current_segment)

    return segments

# Color mapping for consistency across Stage 7 and Stage 8
COLORS = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']

# --- 1. Load the Coastlines ---
segments = load_coastline(IN_DIR / 'world_map.txt')
print(f"Loaded {len(segments)} coastline segments successfully.")
```

## Stage 8: export for Google Earth and write the report

In this final stage, we perform the following exports:
1. **Static 3D Orbit Paths (`orbit_tracks_ecef.kml`)**: Generates continuous absolute coordinate 3D trails of each satellite relative to the rotating Earth.
2. **Animated Ground Tracks (`orbit_animation.kml`)**: Uses Google Earth's time-span extension elements (`gx:MultiTrack` and `gx:Track`) with points sampled every 60 seconds. To prevent visual tracking lines from cutting backwards across the Pacific Ocean on coordinate wraparounds, the tracks are programmatically split whenever the longitude jump exceeds $180^\circ$.
3. **Consolidated Report (`orbit_report.txt`)**: Bundles metadata, model limitations, parsed orbital elements, and computed metrics into a formatted text file.
4. **Downloader Packaging**: Packages both 2D PNG plots, the two KML files, and the report into a single `/content/satellite_orbits/output/satellite_orbit_outputs.zip` file for delivery.

```python
# Ensure all inputs from earlier stages are present in the global scope
if 'elements' not in globals() or 'tracks' not in globals() or 'COLORS' not in globals():
    raise ValueError("Stage 3, 5, or 7 outputs are missing.")

def hex_to_kml_aabbggrr(mpl_color_name):
    """Converts Matplotlib color name to KML aabbggrr hex format."""
    hex_rgba = mcolors.to_hex(mpl_color_name, keep_alpha=False)  # #rrggbb
    r = hex_rgba[1:3]
    g = hex_rgba[3:5]
    b = hex_rgba[5:7]
    return f"ff{b}{g}{r}"  # Force opaque alpha (ff)

# Register namespaces with xml.etree.ElementTree
ET.register_namespace('', "http://www.opengis.net/kml/2.2")
ET.register_namespace('gx', "http://www.google.com/kml/ext/2.2")

# --- 1. Export orbit_tracks_ecef.kml (Static 3D Paths) ---
kml_static = ET.Element('kml', xmlns="http://www.opengis.net/kml/2.2",
                        attrib={"xmlns:gx": "http://www.google.com/kml/ext/2.2"})

doc_static = ET.SubElement(kml_static, 'Document')
name_static = ET.SubElement(doc_static, 'name')
name_static.text = "Multi-Satellite Orbits | Earth-fixed"

for el, track, color in zip(elements, tracks, COLORS):
    kml_color = hex_to_kml_aabbggrr(color)
    style_id = f"style_{el['norad_id']}"

    # Setup unique Placemark styles
    style = ET.SubElement(doc_static, 'Style', id=style_id)
    linestyle = ET.SubElement(style, 'LineStyle')
    ET.SubElement(linestyle, 'color').text = kml_color
    ET.SubElement(linestyle, 'width').text = "2.5"

    iconstyle = ET.SubElement(style, 'IconStyle')
    ET.SubElement(iconstyle, 'color').text = kml_color
    icon = ET.SubElement(iconstyle, 'Icon')
    ET.SubElement(icon, 'href').text = "http://maps.google.com/mapfiles/kml/shapes/placemark_circle.png"

    placemark = ET.SubElement(doc_static, 'Placemark')
    ET.SubElement(placemark, 'name').text = el['name']
    ET.SubElement(placemark, 'styleUrl').text = f"#{style_id}"

    linestring = ET.SubElement(placemark, 'LineString')
    ET.SubElement(linestring, 'extrude').text = "1"
    ET.SubElement(linestring, 'tessellate').text = "1"
    ET.SubElement(linestring, 'altitudeMode').text = "absolute"

    # Accumulate coordinate triplets (longitude, latitude, altitude in meters)
    coord_list = [f"{pt['lon_deg']:.6f},{pt['lat_deg']:.6f},{pt['alt_km']*1000.0:.1f}" for pt in track]
    coord_nodes = ET.SubElement(linestring, 'coordinates')
    coord_nodes.text = " ".join(coord_list)

with open(OUT_DIR / 'orbit_tracks_ecef.kml', 'wb') as f:
    f.write(b'<?xml version="1.0" encoding="UTF-8"?>\n')
    f.write(ET.tostring(kml_static, encoding='utf-8'))

print("Saved 3D tracks KML: orbit_tracks_ecef.kml")
```

```python
# --- 2. Export orbit_animation.kml (Time-Enabled Tracks) ---
kml_anim = ET.Element('kml', xmlns="http://www.opengis.net/kml/2.2",
                      attrib={"xmlns:gx": "http://www.google.com/kml/ext/2.2"})

doc_anim = ET.SubElement(kml_anim, 'Document')
name_anim = ET.SubElement(doc_anim, 'name')
name_anim.text = "Multi-Satellite Ground Tracks | Animation"

for el, track, color in zip(elements, tracks, COLORS):
    kml_color = hex_to_kml_aabbggrr(color)
    style_id = f"anim_style_{el['norad_id']}"

    # Style config
    style = ET.SubElement(doc_anim, 'Style', id=style_id)
    linestyle = ET.SubElement(style, 'LineStyle')
    ET.SubElement(linestyle, 'color').text = kml_color
    ET.SubElement(linestyle, 'width').text = "3"

    iconstyle = ET.SubElement(style, 'IconStyle')
    ET.SubElement(iconstyle, 'color').text = kml_color
    icon = ET.SubElement(iconstyle, 'Icon')
    ET.SubElement(icon, 'href').text = "http://maps.google.com/mapfiles/kml/shapes/placemark_circle.png"

    placemark = ET.SubElement(doc_anim, 'Placemark')
    ET.SubElement(placemark, 'name').text = el['name']
    ET.SubElement(placemark, 'styleUrl').text = f"#{style_id}"

    multitrack = ET.SubElement(placemark, 'gx:MultiTrack')
    ET.SubElement(multitrack, 'altitudeMode').text = "clampToGround"
    ET.SubElement(multitrack, 'gx:interpolate').text = "1"

    # Sample every 60s (using STEP_SECONDS index offsets)
    sample_stride = 60 // STEP_SECONDS
    sampled_pts = [track[idx] for idx in range(0, len(track), sample_stride)]
    if sampled_pts[-1]['time'] != track[-1]['time']:
        sampled_pts.append(track[-1])

    # Split track segments into distinct sub-tracks if longitude wrapping jumps by >180 degrees
    segments = []
    current_seg = []
    for i, pt in enumerate(sampled_pts):
        if i > 0 and abs(pt['lon_deg'] - sampled_pts[i-1]['lon_deg']) > 180.0:
            segments.append(current_seg)
            current_seg = []
        current_seg.append(pt)
    if current_seg:
        segments.append(current_seg)

    # Inject separated gx:Track nodes
    for seg in segments:
        gxtrack = ET.SubElement(multitrack, 'gx:Track')
        for pt in seg:
            # Timestamps must use the UTC pattern: YYYY-MM-DDTHH:MM:SSZ
            t_str = pt['time'].strftime('%Y-%m-%dT%H:%M:%SZ')
            ET.SubElement(gxtrack, 'when').text = t_str
        for pt in seg:
            # Spatial points are formatted as: lon lat altitude
            ET.SubElement(gxtrack, 'gx:coord').text = f"{pt['lon_deg']:.6f} {pt['lat_deg']:.6f} 0"

with open(OUT_DIR / 'orbit_animation.kml', 'wb') as f:
    f.write(b'<?xml version="1.0" encoding="UTF-8"?>\n')
    f.write(ET.tostring(kml_anim, encoding='utf-8'))

print("Saved animated timeline KML: orbit_animation.kml")
```

### Ground Track Analysis and Physics Interpretation

By inspecting our generated plots, we can interpret key orbital dynamics and ground tracks:

* **PULSAR-0 (LEO)**: Demonstrates **multiple dense passes shifting westward**. Because its low altitude means it has an orbital period of only 1.58 hours (much faster than Earth's rotational period), it circles the globe ~15 times a day. While it orbits in an inertial plane, Earth rotates eastward beneath it, shifting successive ground passes westward by approximately $23.7^\circ$ each lap.
* **GSAT0225 (MEO)**: Demonstrates a **wide wave pattern** sweeping across the globe. Since Galileo's semi-synchronous orbit (~14 hours) is comparable to the length of Earth's rotational day, it completes fewer than two ground tracks within our 24-hour cycle, resulting in a gentle, highly extended sine-like wave.
* **SES-5 (GEO)**: Appears as **a single stationary dot** fixed on the equator at exactly $5.0^\circ$ E. Because its orbital period matches Earth's sidereal rotation rate ($23.93$ hours) and its inclination is zero, it orbits in perfect unison with Earth's surface.
* **QZS-2 (HEO)**: Displays a **distinct asymmetric figure-8 shape**. Although it shares the same 24-hour geosynchronous period as SES-5, its $39.2^\circ$ inclination drives its latitude oscillation north and south of the equator. Due to its significant orbital eccentricity ($e \approx 0.075$), it obeys **Kepler's Second Law** (equal areas in equal times), causing it to speed up dynamically near its closest approach (**perigee** at ~32,650 km altitude) and slow down near its furthest point (**apogee** at ~38,935 km altitude). Because it lingers slowly around its northern apogee, it spends much of its orbit hovering over Japan/East Asia, narrowing the top loop of the figure-8 ground track.
